### Install dependencies

```python
import requests
import os 
from dotenv import load_dotenv
import time
import pandas as pd
from sqlalchemy import create_engine, text
from sklearn.preprocessing import MinMaxScaler
```

```python
load_dotenv()

# API key for TMDB API
API_KEY = os.getenv("TMDB_API_KEY")


# Function to get popular movies information
def get_movies_information(oage):

    # Endpoint for popular movies
    url = "https://api.themoviedb.org/3/movie/popular"

    params = {
        "api_key": API_KEY,
        "language": "en-US",
        "page": page
    }

    response = requests.get(url, params=params)

    if response.status_code != 200:
        print(f"Error: {response.status_code}")
        return []

    return response.json()["results"]

all_movies = []

# We will fetch the first 100 pages of popular movies.
for page in range(1, 101):
    print(f"Dowloading page {page}...")
    movies = get_movies_information(page)
    all_movies.extend(movies)

    time.sleep(0.25) # Sleep for 250ms to avoid hitting the rate limit
          
print("Films information: ", len(all_movies))
```

```python
all_movies
```

```python
df = pd.DataFrame(all_movies)
```

```python
df["word_count"] = df["overview"].apply(lambda x: len(str(x).split()))
df
```

```python
# Remove movies with less than 20 words in the overview, as they may not provide enough information for recommendations.
df = df[df["word_count"] > 20]
```

```python
len(df)
```

```python
# To uncrypt the genre ids, we need to make another call to the API.

def get_genres():
    url = "https://api.themoviedb.org/3/genre/movie/list"

    params = {
        "api_key": API_KEY,
        "language": "en-US"
    }

    response = requests.get(url, params=params)

    return response.json()["genres"]

genres = get_genres()

genres
```

```python
# Change the structure of the genres list
new_genres_dict = {genre["id"]: genre["name"] for genre in genres} 
new_genres_dict
```

```python
# Create a new column with the genre names instead of the ids.
df["genres"] = df["genre_ids"].apply(lambda ids: [new_genres_dict.get(i) for i in ids] )
```

```python
df["genres"]
```

```python
# Make sure that the genres columns has no errors
df["errors"] = df["genres"].apply(lambda ids: any(i not in new_genres_dict.values() for i in ids))

# If it is false, there is no erros, if it is true, there is an error.
df[df["errors"] == True]
```

```python
# Check if the genres_columns has any null values or empty lists.
len(df[df["genres"].isnull() | (df["genres"].apply(lambda x: len(x) == 0))])
```

```python
# As there are few rows with empty genres, we can drop them.
df = df[~(df["genres"].isnull() | (df["genres"].apply(lambda x: len(x) == 0)))]
```

```python
important_columns = ['adult', 'id', 'original_language',
       'original_title', 'overview', 'popularity',
       'release_date', 'title', 'vote_average', 'vote_count',
       'genres']
```

```python
df_final = df[important_columns]
df_final
```

```python
df_final.info()
```

```python
# We are going to create a column called "weighted_rating" that will be a weighted average of the vote_average and the popularity of the movie. This will help us to have a better recommendation system.
# This formula computes a weighted rating that balances a movie’s own average rating with the global average rating, giving more influence to movies with a higher number of votes. It reduces the bias of movies with few votes by gradually trusting the global mean when the vote count is low and the movie’s own rating when the vote count is high.
df_final["weighted_rating"] = (
    df_final["vote_count"] / (df_final["vote_count"] + 1000) * df_final["vote_average"]
    + 1000 / (df_final["vote_count"] + 1000) * df_final["vote_average"].mean()
)
```

```python
df_final.columns
```

```python
# We are going to normalize the numeric columns.
# Normalize the numeric columns
# Delete the vote_count column as it is not useful for the recommendation system and it is highly correlated with the weighted_rating column.
df_final2 = df_final.copy()
scaler = MinMaxScaler()
df_final2[["popularity", "vote_average", "weighted_rating"]] = scaler.fit_transform(df_final2[["popularity", "vote_average", "weighted_rating"]])
```

```python
df_final2 = df_final2.drop(columns=["vote_count"])
```

```python
# We are going to create a new column called "content" with the information of the title, overview and genres of the movie.
# We have to apply a lambda function to the genres column to convert the list of genres into a string.
df_final2["genres_text"] = df_final2["genres"].apply(lambda x: " ".join(x))
df_final2["content"] = (
    "id: " + df_final2["id"].astype(str) + ". " +
    "title: " + df_final2["title"].fillna("") + ". " +
    "overview: " + df_final2["overview"].fillna("") + ". " +
    "genres: " + df_final2["genres_text"].fillna("")
)

df_final2
```

```python
# Drop duplicates grouping by id, as it is the primary key of the table.
df_final2 = df_final2.drop_duplicates(subset=["id"])

# Create a column with the year of the release date, as it can be useful for the recommendation system.
df_final2["release_year"] = pd.to_datetime(df_final2["release_date"], errors="coerce").dt.year

df_final2
```

```python
# Cast the release_year to string
df_final2["release_year"] = df_final2["release_year"].astype("Int64")
df_final2
```

```python
df_final2.columns
```

```python
# Drop the column Original title as it is not useful
df_final2 = df_final2.drop(columns = ["original_title"])
df_final2
```

```python
# Data is cleaned so we can use it for the recommendation system. We will save it as a csv file and in a PostgreSQL database.

df_final2.to_csv("movies_information.csv", index=False)
```
