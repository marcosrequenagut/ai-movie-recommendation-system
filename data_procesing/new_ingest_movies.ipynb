#### In this notebook we will upgrade de content column to make it easier to the query of the user, to find a high cosine similarity between a query and a content of a film

```python
import pandas as pd
```

```python
df = pd.read_csv("data\movies_information.csv")
df.head()
```

```python
import ast

# We create a function to create a good content column
def build_content(row):
    
    genres = row["genres"]
    genres = ast.literal_eval(genres)

    g = ", ".join(genres).lower()
    g_and = ", ".join(genres[:-1]) + " and " + genres[-1] if len(genres) > 1 else genres[0]
    g_and = g_and.lower()

    content = (
        f"A {g} movie."
        f"If you like {g} movies watch this. "
        
        f"{row['title']} is a {g_and} film. "

        f"{row['overview']}"
    )

    return content
```

```python
df.drop("content", axis=1, inplace=True)
```

```python
df["content"] = df.apply(build_content, axis=1)
```

```python
import pandas as pd

pd.set_option('display.max_colwidth', None)
print(df["content"])
```

```python
df.to_csv("data/movies_information.csv", index=False)
```
